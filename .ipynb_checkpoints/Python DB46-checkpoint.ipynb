```python
num_string = input("Enter a number: ")
digit_sum = 0
for digit in num_string:
   if digit.isdigit():
       digit_sum += int(digit)
       print(f"The sum of the digits is: {digit_sum}")
```

```python
#Electricity Bill Calculation
#Scenario:
#Up to 100 units → ₹5/unit
#Above 100 units → ₹8/unit

units = int(input("Enter the total units consumed: "))
total_bill=0
for unit in range(1, units + 1):
    if unit <= 100:
        total_bill = unit*5
        print("total_bill=", total_bill)
    
    else:
        total_bill = unit*8
        print("total_bill=", total_bill)

  

       
```

```python
num=int(input("Enter the number of the row:"))
for i in range(num):
    for j in range (num-i):
        print("*", end=" ")
    print()
```

```python
num=int(input("Enter the number of the row:"))
for i in range(num):
    for j in range (i+1):
        print("*", end=" ")
    print()
```

```python
num=int(input("Enter the number of the row:"))
for i in range(num):
    for j in range (i, num):
        print(" ", end=" ")
    for j in range (i+1):
         print("*", end=" ")   
    print()
```
