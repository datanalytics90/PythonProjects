```python
#Exceptional handling -in python exceptions are the less severer error and ity can be handled by try-except-else-finally block
#Syntax
#try:
#   Block of code where erroe may present
#except error_name:
#   Block of codev to handle the given error
#except Exception as e:
#    Block of code to handle the error
#else:
#   return the result if there is no error
#finally:
#   It always run whether the code have any error or not in simple word it provides some information

#Dividing two numbers

try:
  a=int(input("Enter yours 1st number: "))
  b=int(input("Enter your 2nd number: "))
  c=a/b
except ZeroDivisionError:
  print("Do not enter zero to the denominator.")
except Exception as e:      #general error
  print(f"The error is {e}")
else:
  print(c)
finally:
  print("This is the end of the program.")
```
