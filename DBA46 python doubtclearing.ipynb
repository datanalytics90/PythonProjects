```python
lst2 = [1, 2, 3, 2, 4, 1, 2, 5, 3, 1]
frequency = {}

for i in lst2:
    frequency[i] = frequency.get(i, 0) + 1

print(frequency)
```

```python
#Fine the longest word
sentence="Python is widely used in data science field, becauase it is a powerful programming language."
longest_word = ""
for word in sentence.split():
    clean_word = word.strip(",.")

    if len(clean_word) > len(longest_word):
        longest_word = clean_word

print(f"Word: {longest_word}")

```
