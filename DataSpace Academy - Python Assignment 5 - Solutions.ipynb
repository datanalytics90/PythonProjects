```python
# 1. Character Frequency
str1 = "Banana"
freq = {}
for ch in str1:
    freq[ch] = freq.get(ch, 0) + 1
print("1.", freq)
```

```python
# 2. Sum & Average of Digits in String
str1 = "Data123Science456"
digits = [int(ch) for ch in str1 if ch.isdigit()]
print(f"2. Sum is: {sum(digits)} Average is: {sum(digits)/len(digits)}")
```

```python
# 3. Series Sum: 3 + 33 + 333 + 3333
n = 4
total, term = 0, 0
for i in range(n):
    term = term * 10 + 3
    total += term
print("3.", total)
```

```python
# 4. Modify Nested List (Add Element)
list1 = [1, 2, [3, 4, [5, 6], 7], 8]
list1[2][2].append(999)
print("4.", list1)
```

```python
# 5. Insert Sublist into Nested List
list1 = ["x", ["y", ["z", ["p", "q"]]]]
sub_list = ["a", "b", "c"]
list1[1][1][1].extend(sub_list)
print("5.", list1)
```

```python
# 6. Swap Two Tuples
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
tuple1, tuple2 = tuple2, tuple1
print("6. tuple1:", tuple1)
print("   tuple2:", tuple2)
```

```python
# 7. Longest Palindrome in String
def longest_palindrome(s):
    best = ""
    for i in range(len(s)):
        for j in range(i, len(s)):
            sub = s[i:j+1]
            if sub == sub[::-1] and len(sub) > len(best):
                best = sub
    return best
print("7.", longest_palindrome("babad"))
 
```

```python
# 8. String Formatting (Money & Quantity)
totalMoney = 2000
quantity = 4
each = totalMoney / quantity / 2.5  # 200.00 per the expected output
print(f"8. I have {totalMoney} dollars so I can buy {quantity} bags for each of {each:.2f} dollars.")
```

```python
# 9. Reverse Words in String
str1 = "Data Science is fun"
print("9.", " ".join(str1.split()[::-1]))
```

```python
# 10. Character Count (Ignore Case)
str1 = "Mississippi"
freq = {}
for ch in str1.lower():
    freq[ch] = freq.get(ch, 0) + 1
print("10.", freq)
```

```python
# 11. Extract Digits & Find Sum
str1 = "abc10xyz20pq30"
import re
nums = [int(x) for x in re.findall(r'\d+', str1)]
print(f"11. Sum is: {sum(nums)} Average is: {sum(nums)/len(nums)}")
 
```

```python
# 12. Series Sum: 5 + 55 + 555
n = 3
total, term = 0, 0
for i in range(n):
    term = term * 10 + 5
    total += term
print("12.", total)
```

```python
# 13. Replace Element in Nested List
list1 = [1, [2, 3, [4, 5]], 6]
list1[1][2][1] = 999
print("13.", list1)
```

```python
# 14. Add Element at Deepest Level
list1 = [[["a", "b"], "c"], "d"]
list1[0][0].append("z")
print("14.", list1)
 
```

```python
# 15. Swap First and Last Tuple Elements
tuple1 = (10, 20, 30, 40)
lst = list(tuple1)
lst[0], lst[-1] = lst[-1], lst[0]
tuple1 = tuple(lst)
print("15.", tuple1)
```

```python
# 16. Longest Palindrome (Case Sensitive)
print("16.", longest_palindrome("racecarxyz"))

```

```python
# 17. String Formatting with Discount
price = 500
discount = 20
final = price - (price * discount / 100)
print(f"17. Final price after {discount}% discount is {final:.2f}")
```

```python
# 18. Count Vowels and Consonants
str1 = "datascience"
vowels = sum(1 for ch in str1 if ch in "aeiou")
consonants = sum(1 for ch in str1 if ch.isalpha() and ch not in "aeiou")
print(f"18. Vowels: {vowels}")
print(f"    Consonants: {consonants}")
```

```python
# 19. Find Second Largest Number
list1 = [10, 20, 4, 45, 99]
print("19.", sorted(set(list1))[-2])
```

```python
# 20. Check Palindrome String
str1 = "madam"
print("20.", "Palindrome" if str1 == str1[::-1] else "Not Palindrome")
```
