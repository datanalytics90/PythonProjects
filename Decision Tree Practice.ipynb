```python
#import the basic library
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
#import the datasets from scikit learn
#Gives the list of datasets
from sklearn import datasets
print(dir(datasets))
```

```python
#Load the iris data sets
from sklearn.datasets import load_iris
iris=load_iris()
iris
```

```python
#Convert the datasets to dataframe
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
```

```python
df["target"]=iris.target
df.head()
```

```python
#Decide the input and output
x=df.drop("target",axis=1)
y=df["target"]
```

```python
x
```

```python
#['setosa', 'versicolor', 'virginica']   0---->Setosa      1---->Versicolor        2----->Virginica
y
```

```python
#Divide the data sets into training and test data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
#Create the model object
dt_model=DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    random_state=42
)
dt_model
```

```python
#Train the model
dt_model.fit(x_train,y_train)
```

```python
#predict the output
y_pred=dt_model.predict(x_test)
y_pred
```

```python
print(y_test)
```

```python
#Evaluate the model
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
print(f"Accuracy score: {accuracy_score(y_test,y_pred)}")
print(f"Confusion matrix: {confusion_matrix(y_test,y_pred)}")
print(f"Classification report: {classification_report(y_test,y_pred)}")
```

```python
#To visualize the Decision Tree
from sklearn.tree import plot_tree
plt.figure(figsize=(15,10))
plot_tree(dt_model,
          feature_names=iris.feature_names,
          class_names=iris.target_names,
          filled=True,
          rounded=True,
          fontsize=10)
plt.show()
```
