```python
#1. One Hot Encoding(OHE) - Binary value, Nominal data
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

#Create a column
df1=pd.DataFrame({"animal":["Cat","Dog","Camel","Dog","Dog","Camel","Cat"]})
df1
```

```python
#Apply OHE to endode the data
#Create OHE object
ohe=OneHotEncoder()

#Encode the animal column
animal_encoded=ohe.fit_transform(df1[["animal"]])
animal_encoded
```

```python
#Convert the dataset into Data frame
animal_encoded_df=pd.DataFrame(animal_encoded.toarray(),columns=ohe.get_feature_names_out())
animal_encoded_df
```

```python
#2. Label encoding - It assigns numerical values to unique data in a column starting from 0,1,2... and so on
df2=pd.DataFrame({"name":["Rahul","Sanjib","Sachin","Subhashree"],
                  "gender":["Male","Male","Male","Female"],
                  "address":["West Bengsl","West Bengal","Bihar","Odisha"]})
df2
```

```python
#Import label encoder
from sklearn.preprocessing import LabelEncoder
#Create object for Label Encoder
le=LabelEncoder()
le
```

```python
#Encode the column gender and address
df2["gender_encoded"]=le.fit_transform(df2["gender"])
df2["address"]=le.fit_transform(df2["address"])
df2
```

```python
#Ordinal Encoding - It is used for ordinal data
data = {
    "Customer": ["Amit", "Riya", "John", "Sara", "Neha"],
    "Review": ["Good", "Excellent","Average", "Poor", "Good"],
    "Size": ["Medium", "Large", "Small", "Large", "Medium"]
}

df3= pd.DataFrame(data)

print("Original DataFrame:")
print(df3)
```

```python
#Import ordinal encoder
from sklearn.preprocessing import OrdinalEncoder

#Define the order of preference in descending order
categories=[
    ["Poor","Average","Good","Excellent"],
    ["Small","Medium","Large"]
]

#Create the ordinal encoder object
oe=OrdinalEncoder(categories=categories)
oe
```

```python
#Encode the column Review and Size
df3[["review_encoded","size_encoded"]]=oe.fit_transform(df3[["Review","Size"]])
df3
```

```python
#4. Frequency Encoding - While encoding the count of categoriocal data is used for encoding purpose
#Since we are using the count of data for encoding, hence it is also known as count encoding
data = {
    "Customer": ["Amit", "Riya", "John", "Sara", "Neha", "Raj"],
    "City": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai", "Delhi"],
    "Product": ["Laptop", "Mobile", "Laptop", "Tablet", "Mobile", "Laptop"]
}

df4 = pd.DataFrame(data)

print("Original DataFrame:")
print(df4)
```

```python
df4["city_encoded"]=df4["City"].map(df4["City"].value_counts())
df4["product_encoded"]=df4["Product"].map(df4["Product"].value_counts())
df4
```
