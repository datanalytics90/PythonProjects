```python
import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Arjun"],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Education": ["Graduate", "Postgraduate", "Graduate", "School", "Postgraduate"],
    "Experience": [2, 5, 3, 1, 7],
    "Salary": [30000, 55000, 40000, 25000, 70000]
}

df = pd.DataFrame(data)

print(df)
```

```python
#1. Separate the Name column from the dataset.
#2. Apply One-Hot Encoding to the Gender column.
#3. Apply Ordinal Encoding to the Education column using:
#School = 1
#Graduate = 2
#Postgraduate = 3
#4. Apply Standardization to Experience and Salary.
#5. Display the final transformed dataset.
```

```python
#1
df = df.set_index("Name")
df
```

```python
df = pd.get_dummies(df, columns=["Gender"], prefix="Gender", dtype=int)
```

```python
education_map = {"School": 1, "Graduate": 2, "Postgraduate": 3}
df["Education"] = df["Education"].map(education_map)
```

```python
for col in ["Experience", "Salary"]:
    df[col] = (df[col] - df[col].mean()) / df[col].std(ddof=0)
```
