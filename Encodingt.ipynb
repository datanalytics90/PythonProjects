```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
#Create a column
df1=pd.DataFrame({"color":["Red","Blue", "Green" "Orange","Orange","Blue","Red"]})
df1
                 
```

```python
#Apply OHE to encode the data
#Create OHE object
ohe=OneHotEncoder()

#Encode the animal column
color_encoded=ohe.fit_transform(df1[["color"]])
color_encoded
```

```python
#Convert the dataset into Data frame
color_encoded_df1=pd.DataFrame(color_encoded.toarray(),columns=ohe.get_feature_names_out())
color_encoded_df1
```

```python
#Ordinal Encoding - It is used for ordinal data
data = {
    "Customer": ["Amit", "Riya", "John", "Sara", "Neha"],
    "Review": ["Good", "Excellent","Average", "Poor", "Good"],
    "Size": ["Medium", "Large", "Small", "Large", "Medium"]
}

df3= pd.DataFrame(data)

print("Original DataFrame:")
print(df3)
```

```python
#Import ordinal encoder
from sklearn.preprocessing import OrdinalEncoder

#Define the order of preference
categories=[
    ["Poor","Average","Good","Excellent"],
    ["Small","Medium","Large"]
]

#Create the ordinal encoder object
oe=OrdinalEncoder(categories=categories)
oe
```

```python
#Encode the column Review and Size
df3[["review_encoded","size_encoded"]]=oe.fit_transform(df3[["Review","Size"]])
df3
```
