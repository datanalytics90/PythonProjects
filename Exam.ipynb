```python
#Transaction Categorization
#This categorizes transactions into **Low (<1000)**, **Medium (1000–4999)**, and **High (>=5000)**. 
transactions = [1200, 500, 7500, 300, 12000, 4500, 800]

low = []
medium = []
high = []

for t in transactions:
    if t < 1000:
        low.append(t)
    elif t < 5000:
        medium.append(t)
    else:
        high.append(t)

total = 0
print("Low transactions:", low)
print("Medium transactions:", medium)
print("High transactions:", high)
```

```python
#Then computes the total, average, and highest transaction. 
for t in transactions:
    total = total + t

average = total / len(transactions)

highest = transactions[0]
for t in transactions:
    if t > highest:
        highest = t
print("\nTotal amount:", total)
print("Average amount:", round(average, 2))
print("Highest transaction:", highest)
```

```python
#Display the number of transactions in each category.
print("\nCount = Low:", len(low), ", Medium:", len(medium), ", High:", len(high))
```

```python
class Student:
    def __init__(self, name, roll_number, marks1, marks2, marks3):
        self.name = name
        self.roll_number = roll_number
        self.__marks1 = marks1
        self.__marks2 = marks2
        self.__marks3 = marks3

    def calculate_total(self):
        return self.__marks1 + self.__marks2 + self.__marks3

    def calculate_percentage(self):
        total = self.calculate_total()
        return (total / 300) * 100

    def calculate_grade(self):
        percentage = self.calculate_percentage()
        if percentage >= 90:
            return "A+"
        elif percentage >= 80:
            return "A"
        elif percentage >= 70:
            return "B"
        elif percentage >= 60:
            return "C"
        elif percentage >= 50:
            return "D"
        else:
            return "F"

    def display_performance(self):
        print(f"Student Name : {self.name}")
        print(f"Roll Number  : {self.roll_number}")
        print(f"Marks        : {self.__marks1}, {self.__marks2}, {self.__marks3}")
        print(f"Total Marks  : {self.calculate_total()}")
        print(f"Percentage   : {self.calculate_percentage():.2f}%")
        print(f"Grade        : {self.calculate_grade()}")
        print("-" * 35)

# Creating Student Object
student1 = Student("Aditi Sharma", "R101", 85, 90, 78)
student2 = Student("Rohan Verma", "R102", 45, 55, 40)
student3 = Student("Rahul Dutta", "R103", 47, 59, 90)

# Displaying results
student1.display_performance()
student2.display_performance()
student3.display_performance()
```

```python
import numpy as np

sales = np.array([
                  [120, 150, 180],
                  [200, 170, 210],
                  [90, 110, 130],
                  [300, 280, 350]
])
```

```python
# 1. Total sales for each product:
# "axis=1". It sums across columns (months) for each row (product).
product_totals = sales.sum(axis=1)
print("Total sales per product:", product_totals)
```

```python
#2. Total sales for each month:
# axis=0 sums down rows (products) for each column (month).
month_totals = sales.sum(axis=0)
print("Total sales per month:", month_totals)
```

```python
# 3. Identify the product with the highest total sales.
# np.argmax() returns the index of the largest value; 
#+1 converts it to a 1-based product number for display.
highest_index = np.argmax(product_totals)
print("Product with highest sales: Product", highest_index + 1)
print("Total sales:", product_totals[highest_index])
```

```python
# 4. Calculate the overall average sales.
# .mean() with no axis flattens the whole 2D array and averages all 12 values.
overall_average = sales.mean()
print("Overall average sales:", overall_average)
```

```python
# 5. Identify all sales values greater than 200.
high_sales = sales[sales > 200]
print("Sales values greater than 200:", high_sales)
```

```python
#Import Pandas Libery
import pandas as pd

data = {
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales"],
    "Salary": [45000, 38000, 55000, 42000, None, 60000],
    "Experience": [2, 1, 4, 3, 2, 5]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 1. Handle the missing salary value:
# Median is used instead of mean because salary can be skewed by a few high earners, and the median stays closer to a "typical" value.
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print(df)
```

```python
# 2. Average salary for each department:
dept_avg_salary = df.groupby("Department")["Salary"].mean()
print(dept_avg_salary)
```

```python
# 3. Employees with more than 2 years of experience:
experienced = df[df["Experience"] > 2]
print(experienced)
```

```python
# 4. Employee with the highest salary:
# Sorts the whole DataFrame by salary descending, then grabs the first row (.iloc[0]) — the highest earner ends up on top.
highest_paid = df.sort_values("Salary", ascending=False).iloc[0]
print(highest_paid)
```

```python
# 5. Sort the DataFrame by salary in descending order.
sorted_df = df.sort_values("Salary", ascending=False)
print(sorted_df)
```

```python
# EDA & Visualization Task 
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 35, 40, 45, 50, 52, 60],
    "Salary": [25000, 28000, 32000, 35000, 42000, 50000, 58000, 65000, 70000, 90000]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Display descriptive statistics.
print(df.describe())
```

```python
# Check for missing values.
# No missing values in either column.
print(df.isnull().sum())
```

```python
# Identify the correlation between Age and Salary.
# A value this close to +1 indicates a very strong positive linear relationship:
# salary rises almost perfectly in step with age in this dataset.
correlation = df["Age"].corr(df["Salary"])
print("Correlation between Age and Salary:", correlation)
```

```python
# 4 & 5. Scatter plot with title, axis labels, and readability improvements:
# Observation: This confirms the strong positive correlation visually:
# The points trend in an almost straight line as Age increases, matching the 0.99 correlation coefficient from part 3.
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df["Age"], df["Salary"], color="steelblue", s=80, edgecolor="black")

plt.title("Age vs Salary", fontsize=14, fontweight="bold")
plt.xlabel("Age (years)", fontsize=12)
plt.ylabel("Salary (INR)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
