```python
#Given:
#str1 = "Dataspace29@#8496"
#Expected Outcome:
#Sum is: 38 Average is  6.333333333333333

str1 = "Dataspace29@#8496"
digits = [int(char) for char in str1 if char.isdigit()]
total = sum(digits)
average = total / len(digits)
print("Sum is:", total, "Average is ", average)
```

```python
#Write a python code to find the longest palindrome present in any input string
s = input("Enter a string: ")
longest = ""

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):
        sub = s[i:j]
        if sub == sub[::-1] and len(sub) > len(longest):
            longest = sub

print("Longest palindrome:", longest)
```
