```python
import pandas as pd
```

```python
#Load the data
flight=pd.read_excel("/Users/cashify/Desktop/flight_price (1).xlsx")
flight
```

```python
#Change the Object to Datetime
flight["Dep_Time"]=pd. to_datetime(flight["Dep_Time"])
print(flight["Dep_Time"].dtype)
```

```python
# Step 1: Extract hour (already converted above)
flight["Dep_Hour"]=flight["Dep_Time"].dt.hour
print(flight["Dep_Hour"].head(10))
```

```python
#Step 2: Categorize Time Periods
def get_time_period(hour):
    if 0 <= hour < 4:
        return 'Late Night'
    elif 4 <= hour < 8:    
        return 'Early Morning'
    elif 8 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 16:
        return 'Afternoon'
    elif 16 <= hour < 20:
        return 'Evening'
    else:
        return 'Night'
```

```python
#Step 3: Apply udf to the Dataframe
flight["Time Period"] = flight["Dep_Hour"].apply(get_time_period)
```

```python
# Most expensive routes
flight.groupby("Route")["Price"].agg(["mean","count","max","mean"])
```

```python
#Find Top 5 Airlines with Highest Average Ticket Price.
flight.groupby("Airline")["Price"].agg(["mean","count","std"]).sort_values(ascending=False)
```
