```python
import numpy as np
import pandas as pd

import warnings
warnings.filterwarnings("ignore")
```

```python
#Load the flight data
df = pd.read_excel('/Users/cashify/Desktop/flight_price (1).xlsx')
print(df['Dep_Time'].head())
```

```python
#Extract hour from the column "Dep_Time" and create a new column
df['Dep_Hour']=pd. to_datetime(df['Dep_Time']).dt.hour
```

```python
print(df[['Dep_Time', 'Dep_Hour']].dtypes)
```

```python
print(df[['Dep_Time', 'Dep_Hour']].head())
```

```python
#Dep_hour 
#0 to 4 ---> Late Night
#4-8 ----->Early Morning
#8-12 ----->Morning
#12-4 ----> Afternoon
#4-8 ---->Evening
#8-0 ----> Night

def get_time(hour):
    if 0 <= hour < 4:
        return 'Late Night'
    elif 4 <= hour < 8:
        return 'Early Morning'
    elif 8 <= hour < 12:   
        return 'Morning'
    elif 12 <= hour < 16:    
        return 'Afternoon'
    elif 16 <= hour < 20:
        return 'Evening'
    else:
        return'Night'
        
```

```python
#apply() - To apply a UDF to the DataFrame
df["Time_Period"]=df["Dep_Hour"].apply(get_time)
df
df.head(10)
```

```python
#Most Expensive Route
df.groupby(["Route","Airline"])["Price"].sum().sort_values(ascending=False).reset_index().head(10)
```

```python
# Find the top 5 Airlines with Highest Average Ticket Price.

df.groupby("Airline")["Price"].mean().sort_values(ascending=False).reset_index()
```
