```python
#Visualization - It's very easy to analyse or get insight from the data with the help of plot and graphical representation
#In python there are some library available which helps visualization
#python library - Matplotlib, seaborn, plotlt, bokeh

#Matplotlib - It is a python library which helps in making plot for the purpose of analysing data

#import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
#Make a data frame
np.random.seed(42)

sales={
    "month":["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "region":["North","North","North","North","North","North","South","South","South","South","South","South"],
    "sales":np.random.randint(1000,5000,12),
    "profit":np.random.randint(100,500,12),
    "customers":np.random.randint(10,50,12),
    "expenses":np.random.randint(200,400,12)
}

df_2=pd.DataFrame(sales)
df_2
```

```python
#Line plot
plt.figure(figsize=(10,4),facecolor="lightpink")
plt.plot(df_2["month"],df_2["sales"],color="red",linestyle="-",linewidth=2,marker="o")
plt.xlabel("Month",fontweight="bold",fontsize=16)
plt.ylabel("Sales",fontweight="bold",fontsize=16)
plt.title("Month wise sales",fontweight="bold",fontsize=16)
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
#column chart
plt.figure(figsize=(3,3),facecolor="lightblue")
plt.bar(df_2["region"],df_2["sales"],color="pink")
plt.xlabel("Region",fontweight="bold",fontsize=16,color="green")
plt.ylabel("Sales",fontweight="bold",fontsize=16,color="green")
plt.title("Region wise sales",fontweight="bold",fontsize=16,color="red")
plt.show()
```

```python
#Bar plot
plt.figure(figsize=(3,3),facecolor="lightblue")
plt.barh(df_2["region"],df_2["sales"],color="pink")
plt.xlabel("Region",fontweight="bold",fontsize=16,color="green")
plt.ylabel("Sales",fontweight="bold",fontsize=16,color="green")
plt.title("Region wise sales",fontweight="bold",fontsize=16,color="red")


#Channge the background of chart area
plt.gca().patch.set_facecolor("lightblue")
plt.show()
```

```python
#Scatter plot - It is used to get the relationship between two numerical columns
plt.figure(figsize=(5,3),facecolor="lightblue")
plt.scatter(df_2["sales"],df_2["profit"])
plt.xlabel("Sales",fontweight="bold",fontsize=16,color="green")
plt.ylabel("Profit",fontweight="bold",fontsize=16,color="green")
plt.title("Relationship between two numerical columns")


plt.gca().patch.set_facecolor("red")
plt.show()
```

```python
#Multipe line plot
plt.figure(figsize=(8,4))
plt.plot(df_2["month"],df_2["sales"],color="red",linestyle="-",linewidth=2,marker="o")
plt.plot(df_2["month"],df_2["expenses"],color="blue",linestyle="-",linewidth=2,marker="o")
plt.plot(df_2["month"],df_2["profit"],color="green",linestyle="-",linewidth=2,marker="o")
plt.xlabel("Month",fontweight="bold",fontsize=16)
plt.ylabel("Sales",fontweight="bold",fontsize=16)
plt.legend()

plt.show()
```

```python
#Box plot - It is used to check the outliers
plt.figure(figsize=(8,4))
plt.boxplot([df_2["sales"],df_2["expenses"],df_2["profit"]],vert=True,patch_artist=True)
plt.xticks([1,2,3],["sales","expenses","profit"])

plt.show()
```

```python
#Make a pie chart to show the distribution of Region wise sales
region_sales=df_2.groupby("region")["sales"].sum()
region_sales
```

```python
plt.figure(figsize=(4,4))
plt.pie(region_sales,labels=region_sales.index,autopct="%1.0f%%")
plt.show()
```

```python
#Histogram plot - It is used to check the distribution of data
#It is used for continuous numerical data
plt.hist(df_2["sales"],bins=6)
plt.show()
```

```python
#Multiple plots in the single page
plt.figure(figsize=(20,10))

#Month wise sales - plot-1
plt.subplot(2,3,1)
plt.plot(df_2["month"],df_2["sales"],color="red",linestyle="-",linewidth=2,marker="o")
plt.xlabel("Month",fontweight="bold",fontsize=16)
plt.ylabel("Sales",fontweight="bold",fontsize=16)
plt.title("Month wise sales",fontweight="bold",fontsize=16)
plt.xticks(rotation=45)
plt.grid()

#Region wise sales - plot-2
plt.subplot(2,3,2)
plt.bar(df_2["region"],df_2["sales"],color="pink")
plt.xlabel("Region",fontweight="bold",fontsize=16,color="green")
plt.ylabel("Sales",fontweight="bold",fontsize=16,color="green")
plt.title("Region wise sales",fontweight="bold",fontsize=16,color="red")

#Scatter plot - Plot-3
plt.subplot(2,3,3)
plt.scatter(df_2["sales"],df_2["profit"])
plt.xlabel("Sales",fontweight="bold",fontsize=16,color="green")
plt.ylabel("Profit",fontweight="bold",fontsize=16,color="green")
plt.title("Relationship between two numerical columns")

#Multiple line chart - plot-4
plt.subplot(2,3,4)
plt.plot(df_2["month"],df_2["sales"],color="red",linestyle="-",linewidth=2,marker="o")
plt.plot(df_2["month"],df_2["expenses"],color="blue",linestyle="-",linewidth=2,marker="o")
plt.plot(df_2["month"],df_2["profit"],color="green",linestyle="-",linewidth=2,marker="o")
plt.xlabel("Month",fontweight="bold",fontsize=16)
plt.ylabel("Sales",fontweight="bold",fontsize=16)

#Make a pie chart to show the distribution of Region wise sales - Pie chart - plot-5
plt.subplot(2,3,5)
region_sales=df_2.groupby("region")["sales"].sum()

plt.pie(region_sales,labels=region_sales.index,autopct="%1.2f%%")

#Histogram plot - plot-6
plt.subplot(2,3,6)
plt.hist(df_2["sales"],bins=6)

plt.tight_layout()
plt.show()
```

```python
#seaborn - It is a python library which is used for visualization
#It gives much more options to make the chart attractive and presentable

import seaborn as sns

#seaborn library already have a number of in-build data set available for practice
#The available data sets in seaborn are
sns.get_dataset_names()
```

```python
flight=sns.load_dataset("flights")
flight.head()
```

```python
flight["year"].unique()
```

```python
#line plot
plt.figure(figsize=(10,5),facecolor="blue")
sns.lineplot(x="year",y="passengers",data=flight,hue="month")
plt.xlabel("Year",fontweight="bold",fontsize=16)
plt.ylabel("Passengers",fontweight="bold",fontsize=16)
plt.title("Year wise passengers",fontweight="bold",fontsize=16)
plt.gca().patch.set_facecolor("lightblue")
plt.show()
```

```python
tips=sns.load_dataset("tips")
tips.head()
```

```python
plt.figure(figsize=(8,4))
sns.scatterplot(x="total_bill",y="tip",data=tips,hue="sex",style="smoker",palette="bright")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Total Bill vs Tip")
plt.show()
```

```python
#barplot - categorical vs numerical
plt.figure(figsize=(8,4),facecolor="lightblue")
sns.barplot(x="day",y="total_bill",data=tips,hue="sex",palette="cubehelix")
plt.xlabel("Day",color="orange",fontsize=12,fontweight="bold")
plt.ylabel("Total Bill",color="red",fontsize=12,fontweight="bold")
plt.title("Day vs Total Bill")

plt.gca().patch.set_facecolor("pink")
plt.show()
```

```python
#regplot - scatter+regression line
plt.figure(figsize=(8,4),facecolor="blue")
sns.regplot(x="total_bill",y="tip",data=tips)
plt.xlabel("Total Bill",color="orange",fontsize=12,fontweight="bold")
plt.ylabel("Tip",color="red",fontsize=12,fontweight="bold")
plt.title("Total Bill vs Tip")
plt.show()
```

```python
#Pie chart to show the contribution of male and female in the restaurant
gender_count=tips["sex"].value_counts()
gender_count
```

```python
plt.pie(gender_count,labels=gender_count.index,autopct="%1.2f%%")
plt.show()
```

```python
#countplot - it gives a column chart on the basis of count
plt.figure(figsize=(5,3),facecolor="lightpink")
sns.countplot(x="sex",data=tips,hue="smoker")
plt.xlabel("Sex",color="orange",fontsize=12,fontweight="bold")
plt.ylabel("Count",color="red",fontsize=12,fontweight="bold")
plt.title("Count of Male and Female")
plt.show()
```

```python
#box plot - It used to check the outlier
plt.figure(figsize=(8,4),facecolor="lightpink")
sns.boxplot(x="day",y="total_bill",data=tips,hue="smoker")
plt.xlabel("Day",color="orange",fontsize=12,fontweight="bold")
plt.show()
```

```python
#Violin plot - It gives the information of outliers as well as gives the information of distribution of data
plt.figure(figsize=(8,4),facecolor="lightpink")
sns.violinplot(x="day",y="total_bill",data=tips,hue="smoker")
plt.xlabel("Day",color="orange",fontsize=12,fontweight="bold")
plt.show()
```

```python
#Histogram plot - it used to check the distribution of continuous numerical data
plt.figure(figsize=(8,4),facecolor="lightpink")
sns.histplot(x="total_bill",data=tips,bins=25,kde=True)
plt.xlabel("total_bill",color="red",fontsize=12,fontweight="bold")
plt.show()
```

```python
#kde plot -  normal distribution,right skewed distribution or left skewed distribution
plt.figure(figsize=(8,4),facecolor="lightpink")
sns.kdeplot(x="total_bill",data=tips,hue="sex")
plt.xlabel("total_bill",color="red",fontsize=12,fontweight="bold")
plt.show()
```

```python
#Pair plot
plt.figure(figsize=(10,6))
sns.pairplot(data=tips,palette="dark",hue="sex")
plt.show()
```

```python
#Heat map - It gives the correlation between different columns in the data frame
#Calculate the correlation value in the dataset
corr_value=tips.select_dtypes(include="number").corr()
sns.heatmap(corr_value,annot=True,cmap="rocket")
plt.show()
```

```python
#Joint plot
sns.jointplot(x="total_bill",y="tip",data=tips,kind="reg")
plt.show()
```
