```python
#Conditional statements - It is used to perform some condition based operation
#if, if-else, if-elif-else
#if - It handles single condition
#Syntax
#if condition:
#   Block of code
age=int(input("Enter your age: "))
if age>=18:
  print("You are eligible to vote.")
```

```python
#if-else  - It handles two conditions
#Syntax
#if condition:
#   Block of code
#else:
#   Block of code
#The code under else statement will only run when the condition written under if is False.
age=int(input("Enter your age: "))
if age>=18:
  print("You are eligible to vote.")
else:
  print("Go to your school and learn Indian Polity.")
```

```python
#Multiple conditions - if-elif-else
#Syntax
#if condition_1:
#   Block of code
#elif condition_2:
#   Block of code
#elif condition_3:
#   Block of code
#else:
#   Block of code

marks=int(input("Enter your marks: "))
if marks>=90 and marks<=100:
  print("Grade O")
elif marks<90 and marks>=75:
  print("Grade E")
elif marks<75 and marks>=60:
  print("Grade A")
elif marks<60 and marks>=45:
  print("Grade B")
elif marks<45 and marks>=35:
  print("Grade C")
elif marks<0 or marks>100:
  print("Invalid marks")
else:
  print("Grade F")
```

```python
#Python code to find the factorial of any number
#5!=5x4x3x2x1
#n!=1x2x3x...x(n-2)x(n-1)xn
n=int(input("Enter the number: "))
fact=1
if n==0:
  print("Factorial of zero is 1")
elif n<0:
  print("Factorial of negative number doesn't exist.")
else:
  for i in range(1,n+1):
   fact*=i
  print(fact)
```

```python
#armstrong number
#153     number of digits=3     1^3+5^3+3^3=153
num=input("Enter the number: ")
sum=0
n=len(num)
for i in num:
  sum+=int(i)**n        #a+=2    it means a=a+2      sum=sum+int(i)**n
if sum==int(num):
  print(f"The number {num} is an armstrong number.")
else:
  print(f"The number {num} is not an armstrong number.")
```

```python
#Write a python code to get the sum of digits of a number      123---->6
num=int(input("Enter the number: "))
sum=0
while num>0:
  digit=num%10
  sum+=digit
  num//=10
print(f"The sum of the digit is {sum}")
```

```python
#Pattern printing
n=int(input("Enter the number of rows: "))
for i in range(n):
  for j in range(n):
    print("*",end=" ")
  print()
```

```python
#Increasing right angle triangle of *
n=int(input("Enter the number of rows: "))
for i in range(n):
  for j in range(i+1):
    print("*",end=" ")
  print()
```

```python
#Decreasing right angle triangle of *
n=int(input("Enter the number of rows: "))
for i in range(n):
  for j in range(i,n):
    print("*",end=" ")
  print()
```

```python
#Right sided increasing triangle of * - This is a combination of decreasing triangle of space andf increasing triangle of star
n=int(input("Enter the number of rows: "))
for i in range(n):
  for j in range(i,n):
    print(" ",end=" ")
  for j in range(i+1):
    print("*",end=" ")
  print()
```
