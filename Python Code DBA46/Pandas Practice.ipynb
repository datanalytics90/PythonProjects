```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['Rahul', 'Priya', 'Amit', 'Sneha', 'Rohit', 'Neha'],
    'Age': [22, np.nan, 24, 23, np.nan, 25],
    'City': ['Kolkata', 'Delhi', np.nan, 'Mumbai', 'Delhi', np.nan],
    'Salary': [25000, 30000, np.nan, 28000, 35000, np.nan],
    'Department': ['IT', np.nan, 'HR', 'IT', 'Sales', 'HR']
}

print(data)
```

```python
#Display the dataset and check the missing values
df=pd.DataFrame(data)
print("DataFrame is:")
print(df)
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
#Check total missing values
print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
# Check whether any missing values exist
print("\nAre there any missing values?")
print(df.isnull().values.any())
```

```python
#Fill all missing values with 0 using fillna().
df.fillna(0, inplace=True)
print(df.isnull().sum())
```

```python
#Fill missing values in the Age column with the average age.
df['Age']=df['Age'].fillna(df['Age'].mean())
print(df['Age'])
```

```python
#Remove all rows containing missing values using dropna().
df_drop_rows = df.dropna(inplace=True)
print("\nDrop rows with any null:\n", df_drop_rows)
```

```python
#Forward-fill missing values using ffill()
df.ffill()
```

```python
#Backward-fill missing values using bfill().
df.bfill()
```

```python
#Drop the Department column.
df_drop_col= df.drop(columns=['Department'], errors='ignore')
print(df_drop_col)
```
