```python
#Multiple conditions - if-elif-else
#Syntax
#if condition_1:
#   Block of code
#elif condition_2:
#   Block of code
#elif condition_3:
#   Block of code
#else:
#   Block of code

marks=int(input("Enter your marks: "))
if marks>=90 and marks<=100:
  print("Grade O")
elif marks<90 and marks>=75:
  print("Grade E")
elif marks<75 and marks>=60:
  print("Grade A")
elif marks<60 and marks>=45:
  print("Grade B")
elif marks<45 and marks>=35:
  print("Grade C")
elif marks<0 or marks>100:
  print("Invalid marks")
else:
  print("Grade F")
```

```python
#armstrong number
#153     number of digits=3     1^3+5^3+3^3=153
num=input("Enter the number: ")
sum=0
n=len(num)
for i in num:
  sum+=int(i)**n        #a+=2    it means a=a+2      sum=sum+int(i)**n
if sum==int(num):
  print(f"The number {num} is an armstrong number.")
else:
  print(f"The number {num} is not an armstrong number.")
```

```python
#Increasing right angle triangle of *
n=int(input("Enter the number of rows: "))
for i in range(n):
  for j in range(i+1):
    print("*",end=" ")
  print()
```

```python
#Decreasing right angle triangle of *
n=int(input("Enter the number of rows: "))
for i in range(n):
  for j in range(i,n):
    print("*",end=" ")
  print()
```

```python
#Right sided increasing triangle of * - This is a combination of decreasing triangle of space andf increasing triangle of star
n=int(input("Enter the number of rows: "))
for i in range(n):
  for j in range(i,n):
    print(" ",end=" ")
  for j in range(i+1):
    print("*",end=" ")
  print()
```

```python
set_3={1,2,3,4,5,6,7,8,9}
set_4={7,8,9,10,11}

#Union - It combines the two sets
#Syntax        set_1.union(set_2)       or        set_1 | set_2
union_set=set_3.union(set_4)
union_set
```

```python
#Intersection - It returns the common elements from both the set
#Syntax        set_1.intersection(set_2)      or      set_1 & set_2

intersection_set=set_3.intersection(set_4)
print(intersection_set)
```

```python
#Difference of two set - It returns all the elements from first set excluding common elements from both set
#Syntax        set_1.difference(set_2)     or       set_1 - set_2
diff_set=set_3.difference(set_4)
diff_set
```

```python
#Symmetric Difference - Opposite of intersection
#It returns all the elements from both the sets excluding the common elements
#Syntax       set_1.symmetric_difference(set_2)      or     set_1 ^ set_2

sym_diff_set=set_3.symmetric_difference(set_4)
sym_diff_set
```

```python
#Write an udf which returns the factorial of any number
def factorial(n):

  if n<0:
    print("The factorial for negative number do not exist.")
  elif n==0:
    print("The factorial of zero is 1.")
  else:
    fact=1
    for i in range(1,n+1):
      fact*=i
    return fact
```

```python
factorial(5)
```

Electricity charges are:

First 100 units → ₹5/unit

Next 100 units → ₹7/unit

Above 200 units → ₹10/unit

Create a function to calculate total bill.

Create a function to add GST.

Create a function to generate bill details.

```python
def calculate_bill(units):
    if units <= 100:
        return units * 5
    elif units <= 200:
        return 100 * 5 + (units - 100) * 7
    else:
        return units*10
```

```python
calculate_bill(150)
```

```python
#map() - function in python applies a given function to each element of an iterable object(list,set,tuple etc).
#Syntax
#map(function_name,iterable_object)

l1=[1,2,3,4,5]
square=[]
for i in l1:
  square.append(i**2)
print(square)
```

```python
def square(n):
  return n**2

l2=map(square,l1)
print(list(l2))
```

```python
#map+lambda
#Syntax    map(lambda_expression,iterable_object)

l3=map(lambda n:n**2,l1)
l3=list(l3)
print(l3)
```

```python
#Exceptional Handling - Exception as a disruption in the code.
#It is a less severe problem which can be handled by try-except-else-finally block
#Syntax
#try:
#   Block of code where possibility of any error
#except error_name_1:
#   code to handle the error_name_1
#except error_name_2:
#   code to handle the error_name_2
#except Exception as e:
#   print(f"The error is {e}")
#else:
#   this one will only run when the code written under try block is correct
#finally:
#    It always run in simple words it provides information



try:
  a=int(input("Enter Numerator: "))
  b=int(input("Enter denominator: "))
  c=a/b
except ZeroDivisionError:
  print("Enter value for denominator other than 0.")
except ValueError:
  print("Enter intergers/floats to divide.")
except Exception as e:
  print("The error is {e}.")
else:
  print(c)
finally:
  print("The division of two numbers.")
```
