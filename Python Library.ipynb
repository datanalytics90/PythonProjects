```python
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [14, 28, 45, 72]}
df2 = pd.DataFrame(data)
df2
```

```python
#apply() - it helps us to apply udf to the Dataframe
#Syntax       DataFrame["New_col"]=DataFrame["Existing_col"].apply("udf_name")
def categorizer(age):
  if age<=18:
    return "Teen"
  elif age>18 and age<=35:
    return "Young Age"
  elif age>35 and age<=55:
    return "Middle Age"
  else:
    return "Old Age"
```

```python
df2["Age_category"]=df2["Age"].apply(categorizer)
df2
```
