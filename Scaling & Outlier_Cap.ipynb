```python
#Outlier capping - Instead of removing the outlier we can cap the outlier value within the data range
#IQR = Q3-Q1
#upper_fence=Q3+1.5*IQR
#lower_fence=Q1-1.5*IQR

import pandas as pd
data={
    "Name":["Priyanka","Sudeep","Sujoy","Anuska"],
    "Salary":[30000,28000,32000,6768778778778]      #salary contains the outliers
}

df6=pd.DataFrame(data)
df6
```

```python
import numpy as np
Q1=np.percentile(df6["Salary"],25)
Q3=np.percentile(df6["Salary"],75)
IQR=Q3-Q1
upper_fence=Q3+1.5*IQR
lower_fence=Q1-1.5*IQR
print(Q1,Q3,IQR,upper_fence,lower_fence)
```

```python
#cap the salary
df6["Salary"]=np.where(df6["Salary"]>upper_fence,upper_fence,df6["Salary"])
df6["Salary"]=np.where(df6["Salary"]<lower_fence,lower_fence,df6["Salary"])
df6
```

```python
#Option_2
df6["Salary_cap"]=np.clip(df6["Salary"],lower_fence,upper_fence)
df6
```
