```python
# 2. Check missing values and identify columns with the highest missing values.
# Count missing values in every column and sort them from highest to lowest.
# Here, as per my observation, the Rating column has the highest number of missing value.
missing_values = df.isnull().sum().sort_values(ascending=False)
print(missing_values)
df.head()
```

```python
# 3. Check the number of duplicate records in the dataset.
# Count completely duplicated rows present in the dataset.
duplicate_rows = df.duplicated().sum()
print("Number of Duplicate Rows:", duplicate_rows)
```

```python
# 4. Analyze the Category column for unique categories.
# Find the number of unique categories and the five categories containing the most apps.
print("Unique Categories:", df["Category"].nunique())
```

```python
# 4. Analyze the Category column for the top 5 categories.
top_5_categories = df["Category"].value_counts().head(5)
print("\nTop 5 Categories:")
print(top_5_categories)
```
