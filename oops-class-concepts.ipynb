```python
#Python programming are of two types - Procedural Programming and Object Oriented Programming(OOPs)
#OOPs ia a way of organizing code to represent real world things car, students, employee etc.
#Using oops helps us to write the code which is reusable and flexible.
#Concepts cobvered in OOPS
"""1. Class
  2. Object
  3. Inhertitance
  4. Encapsulation
  5. Polymorphism
  6. Abstractions"""
```

```python
#Class - A class is a blueprint or templates for creating any object.
#A class have some properties which is called "ATTRIBUTES" in OOPs and show some behaviours/action which is called "METHODS".
#A class defdines some set of rules which defines any object
#1. Properties of the object - Attributes
#2. Behaviours/Actions - Methods
#We can create any class by using the key word "class"
#class - car         attributes-wheels,color,model       methods-car_engine_start,car_runs,car_stops

class Car:          #Creating a class named car
  wheels=4          #This is an attribute common to all car

  def __init__(self,company,color,model,):
    self.company=company          #Instance attributes
    self.color=color         #Instance attributes
    self.model=model         #Instance attributes
  #Methods of the class car
  def start(self):
    print("Car engine started.")
  def run(self):
    print("Car is running.")
  def stop(self):
    print("The car is stopped.")
  def drive(self):
    print(f"The {self.company} {self.model} of {self.color} colour is driving.")
```

```python
#Object - It the real instances that is made from the class
#If class is the blueprint of the car then object is the actual car whuch is made from the blueprint
#Each object have some unique data based on the class attributes just like car color, car model
#Object is created by calling the class

sanjeeb_car=Car("TATA","Red","Tiago")      #Created an object with the help of class

#Access the attributes of the object - object_name.attributes

print(sanjeeb_car.company)
print(sanjeeb_car.color)
print(sanjeeb_car.model)

#Call the methods of the object -  object_name.methods()

sanjeeb_car.start()
sanjeeb_car.stop()
sanjeeb_car.drive()
```
