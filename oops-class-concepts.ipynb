```python
#Class - A class is a blueprint or templates for creating any object.
#A class have some properties which is called "ATTRIBUTES" in OOPs and show some behaviours/action which is called "METHODS".
#A class defdines some set of rules which defines any object
#1. Properties of the object - Attributes
#2. Behaviours/Actions - Methods
#We can create any class by using the key word "class"
#class - car         attributes-wheels,color,model       methods-car_engine_start,car_runs,car_stops

class Car:          #Creating a class named car
  wheels=4          #This is an attribute common to all car

  def __init__(self,company,color,model,):
    self.company=company          #Instance attributes
    self.color=color         #Instance attributes
    self.model=model         #Instance attributes
  #Methods of the class car
  def start(self):
    print("Car engine started.")
  def run(self):
    print("Car is running.")
  def stop(self):
    print("The car is stopped.")
  def drive(self):
    print(f"The {self.company} {self.model} of {self.color} colour is driving.")
```

```python
#Object - It the real instances that is made from the class
#If class is the blueprint of the car then object is the actual car whuch is made from the blueprint
#Each object have some unique data based on the class attributes just like car color, car model
#Object is created by calling the class

sanjeeb_car=Car("TATA","Red","Tiago")      #Created an object with the help of class

#Access the attributes of the object - object_name.attributes

print(sanjeeb_car.company)
print(sanjeeb_car.color)
print(sanjeeb_car.model)

#Call the methods of the object -  object_name.methods()

sanjeeb_car.start()
sanjeeb_car.stop()
sanjeeb_car.drive()
```

```python
#Encapsulation - It is about protecting the data inside a class. It means wrapping data and the methods operate on the data and restricting direct access of some of the object attributes.
#This prevents accidental changes to your data and hides the sensitive information
#Object attributes can be - public, protected(with single underscore),private(With two underscore)
#Protected - These are those attributes/methods starting with single character and we don't directly access these attributes or methods outside the class
#Private - It's very hard to access

class Employee:
  def __init__(self,name,age,salary):      #attributes defines
    self.name=name           #Public attributes
    self.age=age             #Public attributes
    self.__salary=salary     #Private attributes
```

```python
#Get the private property value
#To access a private attributes we can create a getter method
class Employee:
  def __init__(self,name,age,salary):      #attributes defines
    self.name=name           #Public attributes
    self.age=age             #Public attributes
    self.__salary=salary     #Private attributes
  #Define Getter method to access the private property
  def get_salary(self):
    return self.__salary
```

```python
emp_101.salary    #It throws an error
```

```python
#To get the access of private method take the help of getter method
emp_101.get_salary()
```

```python
#Define the parent class
class Father:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def greet(self):
    print(f"Hello My name is {self.name} and I am {self.age} years old.")

#Define the child class
class Child(Father):
  def __init__(self,name,age,standard,grade):
    super().__init__(name,age)
    self.standard=standard
    self.grade=grade
  def student_info(self):
    return f"My name is {self.name} and I'm in class {self.standard}. I secured {self.grade}."
```

```python
f1=Father("Shyam",32)
f1.greet()
```

```python
c1=Child("Anand",12,5,"B")
print(c1.student_info())
```

```python
#Types of inheritance
#single inheritance -It occurs when a child class inherits from a single parent, allowing it to extend the functionality of the parent class
class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def get_details(self):
    print(f"I'm {self.name} and I'm {self.age} years old.")
class Student(Person):
  def __init__(self,name,age,grade):   #an extra attributes grade is created
    #super().__init__(name,age)
    self.name=name
    self.age=age
    self.grade=grade

  def get_student_details(self):
    print(f"I'm {self.name}.\nI'm {self.age} tears old and currently in grade {self.grade}.")
```

```python
s1=Student("Sachin",27,"Msc")
s1.get_student_details()
```

```python
#Multiple inheritance
#Here a child class inherit attributes and methods from multiple parents
class Father:
  def details_father(self):
    print("This is the properties of father")
class Mother:
  def details_mother(self):
    print("This is the properties of mother")
class Child(Father,Mother):    #Here the child class inherit properties from both mother and father
  def details_child(self):
    print("This is the properties of child combining father and mother property and including hi/her own properties.")
```

```python
c1=Child()
c1.details_father()
c1.details_mother()
c1.details_child()
```

```python
#Multilevel inheritance
class GrandFather:
  def details_grand_father(self):
    print("This is the properties of Grand father")
class Father(GrandFather):
  def details_father(self):
    print("This is the properties of father")
class Child(Father):    #Here the child class inherit properties from both mother and father
  def details_child(self):
    print("This is the properties of child combining father and grand father's property and including hi/her own properties.")
```

```python
#Hierarchical inheritance
#Multiple child classes access the properties and methods from single child class.
class Parent:
  def deatils_parent(self):
    print("This the properties of parent.")
class Son(Parent):
  def details_son(self):
    print("This is the properties of son and half of the properties of Parents.")
class Daughter(Parent):
  def details_daughter(self):
    print("This is the properties of daughter and half of the properties of Parents.")
```
